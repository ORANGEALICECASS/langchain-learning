```python
from google.colab import userdata
from pathlib import Path
from dotenv import set_key, load_dotenv

# Fichier .env déjà créé
env_path = Path("/content/lca-lc-foundations/.env")

# Récupérer les clés depuis Colab Secrets
api_keys = [
    "OPENAI_API_KEY",
    "TAVILY_API_KEY",
    "GOOGLE_API_KEY",
    "ANTHROPIC_API_KEY",
    "LANGSMITH_API_KEY"
]

if not env_path.is_file():
    print("Fichier .env introuvable. Vérifie son emplacement.")
else:
    for key in api_keys:
        try:
            value = userdata.get(key)
            if value:
                set_key(str(env_path), key, value)
                print(f"{key} : ajoutée au .env")
            else:
                print(f"{key} : secret vide")
        except Exception:
            print(f"{key} : secret non disponible")

    # Configuration LangSmith
    set_key(str(env_path), "LANGSMITH_TRACING", "true")
    set_key(str(env_path), "LANGSMITH_PROJECT", "lca-lc-foundation")

    # Charger toutes les variables du .env
    load_dotenv(env_path, override=True)

    print("\nConfiguration .env terminée !")
    print("LangSmith Tracing activé.")
```

## Initialising and invoking a model

```python
from langchain.chat_models import init_chat_model

model = init_chat_model(model="gpt-5-nano", use_responses_api=True)
```

```python
response = model.invoke("What's the capital of the Moon?")

response
```

```python
print(response.content)
```

```python
from pprint import pprint

pprint(response.response_metadata)
```

## Customising your Model

```python
model = init_chat_model(
    model="gpt-5-nano",
    use_responses_api=True,
    # Kwargs passed to the model:
    temperature=1.0
)

response = model.invoke("What's the capital of the Moon?")
print(response.content)
```

## Initialising and invoking an agent

```python
from langchain.agents import create_agent

agent = create_agent(model=model)
```

```python
agent = create_agent(model="claude-sonnet-4-6")
```

```python
agent = create_agent("gpt-5-nano")
```

```python
from langchain.messages import HumanMessage

response = agent.invoke(
    {"messages": [HumanMessage(content="What's the capital of the Moon?")]}
)
```

```python
from pprint import pprint

pprint(response)
```

```python
print(response['messages'][-1].content)
```

```python
from langchain.messages import AIMessage

response = agent.invoke(
    {"messages": [HumanMessage(content="What's the capital of the Moon?"),
    AIMessage(content="The capital of the Moon is Luna City."),
    HumanMessage(content="Interesting, tell me more about Luna City")]}
)

pprint(response)
```

## Streaming Output

```python
for token, metadata in agent.stream(
    {"messages": [HumanMessage(content="Tell me all about Luna City, the capital of the Moon")]},
    stream_mode="messages"
):

    # token is a message chunk with token content
    # metadata contains which node produced the token

    if token.content:  # Check if there's actual content
        print(token.content, end="", flush=True)  # Print token
```
